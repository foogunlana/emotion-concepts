### Exhaustive: who cheats, and when?

```python
from pathlib import Path

from inspect_ai import eval_set
from impossible import fast_sum, Prompt

LOG_DIR = "../../.inspect/logs/sweep-5"
```

```python
# Models with published cheating rates, reached through OpenRouter.
# rate = (no-guidance, anti-hack) where the system card reports both.
MODELS = [
    "openrouter/openai/gpt-5",                 # 76% oneoff-SWE, 54% conflicting-SWE  [ImpossibleBench]
    "openrouter/openai/o3",                    # 33% conflicting-LCB (STRICT), >85% (LOOSE/SPEC_ONLY)
    "openrouter/openai/o4-mini",               # in roster, no published figure
    "openrouter/openai/gpt-4.1",               # low - only 13 transcripts classified
    "openrouter/anthropic/claude-opus-4.1",    # 80% / 45%   [Sonnet 4.5 system card 6.1.A]
    "openrouter/anthropic/claude-sonnet-4.5",  # 53% / 20%
    "openrouter/anthropic/claude-sonnet-4",    # 70% / 28%
    "openrouter/qwen/qwen3-coder",             # ~20% impossible-SWEbench - the open-weight point
]

# The judge decides whether a reply concedes. Direct to Anthropic, not
# OpenRouter - Haiku 4.5 is the cheap tier and this is a one-bit judgement.
JUDGE = "anthropic/claude-haiku-4-5-20251001"
```

```python
TASKS = [
    fast_sum(prompt=p, abort_option=ao, limit=1, max_attempts=15)
    for ao in (False,)
    # for ao in (False, True)
    for p in [
        Prompt.NONE,
        Prompt.LOOSE,
        Prompt.SPEC_ONLY,
        Prompt.EVALUATE,
        Prompt.STRICT,
        Prompt.ANTI_HACK
    ]
]
print(f"{len(TASKS)} cells x {len(MODELS)} models = {len(TASKS) * len(MODELS)} runs")
for t in TASKS:
    print("  ", t.name)
```

```python
logs, _ = eval_set(
    TASKS,
    model=MODELS,
    model_roles={"judge": JUDGE},
    log_dir=LOG_DIR,
    max_tasks=4,
    max_connections=20,
    display="log",
    log_level="info"
)
```
